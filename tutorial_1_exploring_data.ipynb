## 9. Building our working dataset

From here on we'll only look at each patient's **baseline** (first) visit — using every visit for every patient would let the same person show up many times, which will matter once we build models next session.

````python
data_reduced = data[data['VISCODE'] == 'bl'].copy()   # .copy() so we can safely edit it
data_reduced.shape
````

### 9.1 Dropping columns we don't need

`.drop()` removes rows (`axis=0`, default) or columns (`axis=1`) by label.

````python
data_reduced.drop('RID', axis=1).head(3)                  # drop one column
data_reduced.drop(['RID', 'VISCODE'], axis=1).head(3)      # drop several
````

> **Exercise**: dropping columns by *pattern*
>
> We want to remove every column whose name contains `_bl` or `Ecog` (baseline-duplicate and questionnaire sub-scores we won't use). Walk through this loop:
>
> - start an empty list `column_names_unwanted`
> - go through every column name
> - if the name contains `'_bl'` or `'Ecog'`, add it to the list

````python
column_names_unwanted = []

for column_name in data_reduced.columns:
    if ('_bl' in column_name) or ('Ecog' in column_name):
        column_names_unwanted.append(column_name)

column_names_unwanted
````

Now drop them, and also the stray `Unnamed: 109` column from earlier, overwriting `data_reduced`:

````python
data_reduced = data_reduced.drop(column_names_unwanted, axis=1).copy()
data_reduced = data_reduced.drop('Unnamed: 109', axis=1).copy()
data_reduced.shape
````

### 9.2 Missing values

We care most about three columns for what's coming next: diagnosis (`DX`), the cognitive test score (`MMSE`), and hippocampus volume (`Hippocampus`). Let's check them for missing values.

````python
data_reduced['DX'].unique()   # categorical: check with unique()
````

````python
data_reduced['MMSE'].isnull().values.any()          # numeric: check with isnull()
data_reduced['Hippocampus'].isnull().values.any()
````

> **Interpretation exercise**
>
> `dropna(axis=0, subset=[...])` drops rows missing a value in any of the listed columns. How many rows would we have left if we dropped rows missing `DX`, `MMSE`, or `Hippocampus`? Is that a lot of data to lose?

````python
print("Rows before:", data_reduced.shape[0])
print("Rows after:", data_reduced.dropna(axis=0, subset=['DX', 'MMSE', 'Hippocampus']).shape[0])
````

## 10. Visualizing the data

Numbers in a table only tell you so much — plots make patterns visible immediately.

> *"Graphical excellence is that which gives to the viewer the greatest number of ideas in the shortest time with the least ink in the smallest space."* — Edward Tufte

````python
import matplotlib.pyplot as plt

ad = pd.read_csv('data/alzheimers_disease_reduced.csv')
data = ad[ad['VISCODE'] == 'bl']   # (already filtered, but never trust your data blindly!)
````

### 10.1 Histogram — distribution of a single variable

Let's look at the age distribution.

````python
data['AGE'].describe()
````

Wait — the **minimum age is 0**. That's clearly not a real patient age; it's a data artifact. This is exactly the kind of thing `.describe()` is good for catching *before* you plot or model anything.

````python
print("Number of rows with AGE == 0:", (data['AGE'] == 0).sum())
print("Smallest age otherwise:", data[data['AGE'] != 0]['AGE'].min())
print("Missing ages:", data['AGE'].isnull().sum())
````

We'll leave these rows in for now (there aren't many), but keep this in mind: **always sanity-check your data before you trust a plot of it.**

````python
fig, ax = plt.subplots(1, 1)
ax.hist(data['AGE'].dropna(), bins=20, color='darkgreen')
ax.set_xlabel('Age')
ax.set_ylabel('Count')
ax.set_title('Distribution of Age')
plt.show()
````

### 10.2 Boxplot — comparing groups

Boxplots show the median, the middle 50% (box), and the range/outliers (whiskers) — great for comparing a numeric variable across categories. Note: unlike `.hist()`, `.boxplot()` doesn't drop `NaN`s for you, so we remove them manually first.

````python
nonna_age = data[~data['AGE'].isnull()]
data_f = nonna_age[nonna_age['PTGENDER'] == 'Female']
data_m = nonna_age[nonna_age['PTGENDER'] == 'Male']

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(8, 4))

ax1.boxplot([data_f['AGE'], data_m['AGE']], labels=['Female', 'Male'])
ax1.set_ylabel("Age")
ax1.set_title("Patients' age")

ax2.boxplot([data_f['PTEDUCAT'], data_m['PTEDUCAT']], labels=['Female', 'Male'])
ax2.set_ylabel("Years of education")
ax2.set_title("Patients' education level")

plt.show()
````

### 10.3 Seaborn: the same ideas, faster

*Seaborn* is built on top of *matplotlib*, with nicer defaults and tight integration with pandas — you mostly just point it at a DataFrame and column names.

````python
import seaborn as sns
sns.set_theme()
````

````python
g = sns.histplot(data=data, x='AGE', bins=20)
g.set(xlabel='Age', title='Distribution of Age (Seaborn)');
````

````python
g = sns.countplot(data=data, x='DX', hue='PTGENDER')
g.set(xlabel='Diagnosis')
g.legend(title='Gender');
````

### 10.4 Relationships between variables

A scatter (or "joint") plot shows the relationship between two numeric variables at once — here, age vs. whole-brain volume, colored by diagnosis.

````python
plt.figure(figsize=(8, 5))
colors = ('blue', 'red', 'darkorange')
g = sns.scatterplot(data=data, x='AGE', y='WholeBrain', hue='DX', palette=colors, alpha=0.4)
g.set(xlabel='Age', ylabel='Brain volume')
legend = g.legend()
legend.texts[0].set_text("Diagnosis")
plt.show()
````
